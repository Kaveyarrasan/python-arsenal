```python
def menu_items():
    """Display menu items and their costs."""
    food_item_list = ['Pizza', 'Burger', 'Pasta', 'Sandwich', 'Salad']
    food_item_cost = {
        food_item_list[0]: 150,
        food_item_list[1]: 120,
        food_item_list[2]: 180,
        food_item_list[3]: 100,
        food_item_list[4]: 80
    }

    print("-"*10)
    print("Menu Card:")
    print("-"*10)
    for item, cost in food_item_cost.items():
        print(f"{item}: ₹{cost}/-")

menu_items()
```

```python
a = 0.01
b = 0.0

print(f"{a or 'Default Value'} | type: {type(a or 'Default Value')}")  # Output: 0.01 | type: <class 'float'>
print(f"{b or 'Default Value'} | type: {type(b or 'Default Value')}")  # Output: Default Value | type: <class 'str'>
```
