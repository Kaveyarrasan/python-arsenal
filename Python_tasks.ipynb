a. Write a program that takes marks as input (initially as a string). \

 - Marks >= 90 --> Outstanding

D. If the input is not numeric, print: \
 Invalid marks entered — Please provide numeric input.


```python
subjects = ['Tamil', 'English', 'Maths', 'Science']
total_mark = 0

try:
    for subject in subjects:
        msg = f"Enter {subject} mark: "
        subject_mark = input(msg)
        total_mark += int(subject_mark)

    total_mark /= len(subjects)
    total_mark = round(total_mark)
    print(f"Total Mark: {total_mark}")

    if total_mark >= 90:
        print("Outstanding")
    elif total_mark >= 75:
        print("Excellent")
    elif total_mark >= 50:
        print("Pass")
    else:
        print("Fail")
except Exception:
    print("Invalid marks entered — Please provide numeric input.")
```

- Check room availability

        - If guest is VIP
            → Offer complimentary upgrade
        - Else if member 5+ years

            → Standard price
    - Else:
        → Show: "No rooms available"


```python
try:
    available_rooms = 5   # Assuming fixed no. of rooms available
    room_price = 1500
    discount = 20
    vip_offer = 500
    vip_room_price = 3000

    room_enquiry = int(input("Enter no. of rooms required (2 heads per room): "))

    if room_enquiry <= available_rooms:
        is_guest_vip = input("Are you a premium card holder (yes/no): ")
        if is_guest_vip.lower() == "yes":
            vip_room_stay_cost = (vip_room_price * room_enquiry)
            vip_complementary_offer = vip_room_stay_cost - vip_offer
            print("Cost price of", str(room_enquiry) ,"premium room(s): " + str(vip_room_stay_cost))
            print("Complementary offer applied to VIP guest and finalized cost: " + str(vip_complementary_offer))
        else:
            is_regular_member = input("Do you own a membership card (5+years) (yes/no): ")
            if is_regular_member.lower() == "yes":
                regular_room_stay_cost = room_price * room_enquiry
                room_discount = (regular_room_stay_cost * discount) / 100
                regular_room_discount_cost = regular_room_stay_cost - room_discount
                print("Cost price of", str(room_enquiry) ,"room(s): " + str(regular_room_stay_cost))
                print("Discounted price: " + str(regular_room_discount_cost))
            else:
                regular_room_stay_cost = room_price * room_enquiry
                print("Standard price of", str(room_enquiry) ,"room(s): " + str(regular_room_stay_cost))
    else:
        print("No rooms available")
except Exception as e:
    print(e)
```

### **J. Looping Constructs**

**Use Case 1: Table Generator** \
Write a program that takes a number from the user and prints the multiplication table from 1 to 10 for that number.

Example: \
 If user enters 5, output should be:

     5 x 1 = 5
     5 x 2 = 10
     ...
     5 x 10 = 50


```python
try:
    multiplicand = int(input("Enter multiplicand value: "))

    multiplier = 1  # multiplier begins with 1
    while multiplier <= 10:   # multiplie iterates till 10
        product = multiplicand * multiplier
        print(f"{multiplicand} x {multiplier} = {product}")
        multiplier += 1
except Exception as e:
    print("Invalid multiplicand: Please enter a number")
```

**Use Case 2: Sum of Even and Odd Numbers**

Write a program that asks the user for a positive integer n. \
Using a loop, calculate and print:
- Sum of all even numbers from 1 to n
- Sum of all odd numbers from 1 to n


```python
try:
    number = int(input("Enter a positive number: "))
    i = 1
    even_number, odd_number = 0, 0
    if number > 0:
        while i <= number:
            if i % 2 == 0:
                even_number += i
                # print(i, "| even_number:", even_number)
            else:
                odd_number += i
                # print(i, "| odd_number:", odd_number)
            i += 1
        print(f"Sum of all even numbers from 1 to {number}: {even_number}")
        print(f"Sum of all odd numbers from 1 to {number}: {odd_number}")
    else:
        raise Exception("Invalid number")
except Exception:
    print("Invalid number")
```

**Use Case 3: Table Generator - Nested Looping**

Write a program that takes a list of values/(number and generate a range) from the user and prints the multiplication table from 1 to 10 for that list of numbers.

Example: \
 If user enters [2,3,5], output should be:

    2 x 1 = 2
    2 x 2 = 4
    ….
    3 x 1 = 3
    3 x 2 = 6
     5 x 1 = 5
     5 x 2 = 10
     ...
     5 x 10 = 50


```python
try:
    prompt_itr_range = int(input("Enter a number to raise prompt iterating upto given number: "))
    i = 1
    number_list = []
    while i <= prompt_itr_range:
        msg = f"Enter a multiplicand to generate multiplication tables: Prompt remaining({(prompt_itr_range-  i)})"
        number_list.append(int(input(msg)))
        i += 1
    print("Entered number list:",number_list)
    print("-"*10)
    for multiplicand in number_list:
        multiplier = 1
        while multiplier <= 10:
            product = multiplicand * multiplier
            print(f"{multiplicand} x {multiplier} = {product}")
            multiplier += 1
        print("-"*10)
except Exception as e:
    print(e)
```
