**Use Case 2:** \
Demonstrate dynamic inference, dynamic typing using with fee by applying .18 gst  and prove strongly typing character also by operating it with Eighteen percent gst

    fee = 45000


```python
# Dynamic Inferencing
fee = 45000
gst = 0.18
print(f"Fee: {fee} | type(fee): {type(fee)}")
print(f"GST: {gst} | type(gst): {type(gst)}")

applied_gst = fee * gst
print(f"Applied GST: {applied_gst} | type(applied_gst): {type(applied_gst)}")
```

```python
# Dynamic Typing
fee = "45000"
gst = 0.18
print(f"Fee: {fee} | type(fee): {type(fee)}")
print(f"GST: {gst} | type(gst): {type(gst)}")

# Strongly Typing
try:
    applied_gst = fee * gst
    print(f"Applied GST: {applied_gst} | type(applied_gst): {type(applied_gst)}")
    raise TypeError()
except TypeError as err_message:
    print("Strongly Typed Error: Can't multiply fee with gst as fee is string type.")
    print("Strongly Typed Error:", str(err_message).capitalize())
```

**Use Case 2: Student Result Classification**

a. Write a program that takes marks as input (initially as a string). \
B. Check if the value can be converted to float. \
C. Then classify (try using if condition with the help of AI, however we will learn about if condition soon):
 - Marks >= 90 --> Outstanding
 - Marks >= 75 --> Excellent
 - Marks >= 50 --> Pass
 - Marks < 50 --> Fail

D. If the input is not numeric, print: \
 Invalid marks entered — Please provide numeric input.


```python
subjects = ['Tamil', 'English', 'Maths', 'Science']
total_mark = 0

try:
    for subject in subjects:
        msg = f"Enter {subject} mark: "
        subject_mark = input(msg)
        total_mark += int(subject_mark)

    total_mark /= len(subjects)
    total_mark = round(total_mark)
    print(f"Total Mark: {total_mark}")

    if total_mark >= 90:
        print("Outstanding")
    elif total_mark >= 75:
        print("Excellent")
    elif total_mark >= 50:
        print("Pass")
    else:
        print("Fail")
except Exception:
    print("Invalid marks entered — Please provide numeric input.")
```

### **H. Python Operators Usecases**

**Use Case 1: Internet Data Usage Calculator**

Write a program that asks the user for:

- Total monthly data limit (in GB)
- Data used so far (in GB)


***Calculate using arithmetic operators:***

    Remaining data = limit - used
    Usage percentage = (used / limit) * 100

***Print:***
- Remaining data
- Usage percentage rounded to 2 decimals


If usage percentage is greater than or equal to 80, print: \
 "Warning: High usage, consider upgrading your plan."


```python
limit = int(input("Enter total monthly data limit (in GB): "))
used = int(input("Enter data used so far (in GB): "))

remaining_data = limit - used
usage_percent = round((used / limit) * 100, 2)

print("Remaining data: " + str(remaining_data))
print("Usage Percentage: " + str(usage_percent) + "%")

if usage_percent >= 80:
    print("Warning: High usage, consider upgrading your plan.")
```
