**Use Case 2: Salary Processing System (HR / Payroll Automation)**

Demonstrates input/output functions, default arguments, arbitrary args, returning values, and composing multiple functions.
    
    What Problem It Solves:
    Companies calculate employee salary differently depending on bonus percentage, incentives, PF, tax, etc.

    How Function-Based Programming Helps:
    We create multiple small reusable functions (bonus calc, tax calc, net salary calc)
    Then compose them together.

```python
# We create multiple small reusable functions (bonus calc, tax calc, net salary calc)

# bonus calculation function
def bonus_calc(salary, bonus_percentage):
    return salary * bonus_percentage  # calculating bonus with salary

# tax calculation function as per new tax slabs in India following new regime for FY 2026-27
def tax_calc(salary):
    tax_value = 0

    if salary <= 1200000:
        return tax_value
        
    elif salary > 1200000:
        if 400001 <= salary <= 800000:
            tax_value = (salary - 400000) * 0.05
            
        elif 800001 <= salary <= 1200000:
            tax_value = (400000 * 0.05) + (salary - 800000) * 0.10
            
        elif 1200001 <= salary <= 1600000:
            tax_value = (400000 * 0.05) + (400000 * 0.10) + (salary - 1200000) * 0.15
            
        elif 1600001 <= salary <= 2000000:
            tax_value = (400000 * 0.05) + (400000 * 0.10) + (400000 * 0.15) + (salary - 1600000) * 0.20
            
        elif 2000001 <= 400000 <= 2400000:
            tax_value = (400000 * 0.05) + (400000 * 0.10) + (400000 * 0.15) + (400000 * 0.20) + (salary - 2000000) * 0.25
            
        else:
            tax_value = (400000 * 0.05) + (400000 * 0.10) + (400000 * 0.15) + (400000 * 0.20) + (400000 * 0.25) + (salary - 2400000) * 0.30
        
        tax_value += tax_value * 0.04
        return tax_value
    return None


# net salary calculation function
def net_salary_calc(salary, bonus_percentage, incentives):
    basic_salary = salary * 0.50  # Assuming basic salary is 50% of total salary
    dearness_allowance = salary * 0.20  # Assuming DA is 20% of total salary
    pf = (basic_salary + dearness_allowance) * 0.12  # Assuming PF is 12% of basic salary + dearness allowance

    bonus = bonus_calc(salary, bonus_percentage)  # calculating bonus with salary and bonus percentage
    tax = tax_calc(salary)   # calculating tax with salary

    # calculating net salary from salary, bonus, incentives and other deductions like tax and pf
    net_salary = salary + bonus + incentives - (tax + pf)
    
    return net_salary
```

```python
print(net_salary_calc(2500000, 0.10, 50000))  # Example usage of net salary calculation
```

**Usecase3: Number Utility Tool**

Create functions:
- is_even(num) → returns True/False
- find_max(*numbers) → returns highest number from arguments
- perform_operation(num, operation) →
    1. If operation = "square" → return num*num
    2. If operation = "cube" → return num*num*num
    3. If unknown → return "Invalid operation"

Test the functions with at least 5 numbers.

```python
def is_even(num):
    """Check if a number is even."""
    # global val
    val = False
    try:

        val = num % 2 == 0
    except (ValueError, TypeError, SyntaxError, NameError):
        print(f"Invalid input: {num} is not a number. Please provide a valid integer.")
    except Exception as e:
        print(f"Invalid input: {e}. Please provide a valid integer.")
        
    return val
```

```python
is_even(4)  # Example usage of is_even function
```

```python
# Example usage of is_even function
print(f"is number 4 even? {is_even(4)}")
print(f"is number 1 even? {is_even(1)}")
print(f"is number a even? {is_even('a')}")
print(f"is number . even? {is_even('.')}")

# Flagging Syntax Error before entering function: is_even function
# print(f"is number a even? {is_even(a)}")
# print(f"is number . even? {is_even(.)}")
```

```python
# find_max(*numbers) → returns highest number from arguments

def find_max(*numbers):
    """Find the maximum number from the provided arguments."""
    return max(*numbers)
```

```python
# Example usage of find_max function
print(type(find_max(100, 201, 186, 424, 125)), type((100, 201, 186, 424, 125)), sep=" | ")
print(type(find_max({1, 2, 7, 4, 5})), type({1, 2, 7, 4, 5}), sep=" | ")
print(type(find_max([1, 2, 8, 4, 5])), type([1, 2, 8, 4, 5]), sep=" | ")

print(f"Maximum number from {type((100, 201, 186, 424, 125))} list (100, 201, 186, 424, 125): {find_max(100, 201, 186, 424, 125)}")
print(f"Maximum number from {type((-1000, 20, 6, 4, 15))} list (-1000, 20, 6, 4, 15): {find_max(-1000, 20, 6, 4, 15)}")
print(f"Maximum number from {type((1, 2, 6, 4, 5))} list (1, 2, 6, 4, 5): {find_max(1, 2, 6, 4, 5)}")
print(f"Maximum number from {type(({1, 2, 7, 4, 5}))} list {{1, 2, 7, 4, 5}}: {find_max({1, 2, 7, 4, 5})}")
print(f"Maximum number from {type(([1, 2, 8, 4, 5]))} list [1, 2, 8, 4, 5]: {find_max([1, 2, 8, 4, 5])}")
```

```python
""" 
- perform_operation(num, operation) →
    1. If operation = "square" → return num*num
    2. If operation = "cube" → return num*num*num
    3. If unknown → return "Invalid operation"
"""

def perform_operation(num, operation:str):
    """Perform arithmetic operations such as square or cube."""
    try:
        if operation.lower() == "square":
            return num*num
        elif operation.lower() == "cube":
            return num*num*num
        else:
            # print("Invalid Operation")
            raise TypeError("Invalid operation")
    except Exception as e:
        # print("Invalid Operation")
        return f"Error: {e}"
```

```python
# Example usage of perform_operation function
print(f"Square of 5: {perform_operation(5, 'square')}")
print(f"Cube of 3: {perform_operation(3, 'cube')}")
print(f"SQUARE of 5: {perform_operation(5, 'SQUARE')}")
print(f"CUBE of 3: {perform_operation(3, 'CUBE')}")
print(f"CUBE of 3: {perform_operation('3', 'CUBE')}")
print(f"CUBE of 5: {perform_operation('5', 'UBE')}")
print(f"No operation on 4: {perform_operation(4, 'None')}")

# print(f"CUBE of 5: {perform_operation(5, UBE)}")   # NameError: name 'UBE' is not defined.
```

**Use Case 5: Invoice Generator (\*\*kwargs)**

Create a function generate_invoice(**products).  
Each key is a product name and each value is its price.  
Print all product names with prices and also print the total amount.


```python
def generate_invoice(**kwargs):
    item = ''
    try:
        total_amount = 0
        for item, cost in kwargs.items():
            item = str(item).capitalize()
            cost = float(cost) if cost >= 0 else 0.0
            print(f"{item} = ₹{cost}/-")
            total_amount += cost
    except Exception:
        print(f"Error: Invalid cost value provided to {item}.\nDetails: String values are not allowed. Please provide a valid numeric cost.")
        return 0

    return f"Total Invoice Amount = ₹{total_amount}/-"
```

```python
generate_invoice(milk=50, egg=60, oil='sd')  # Example usage of generate_invoice function
```

```python
generate_invoice(milk=50, egg=60, biscuit=30, bread=40, jam=80)  # Example usage of generate_invoice function
```

```python
def generate_invoice_type_1(**kwargs):
    # keeping only the necessary requisites
    milk = kwargs.get('milk', 0)
    egg = kwargs.get('egg', 0)
    biscuit = kwargs.get('biscuit', 0)
    bread = kwargs.get('bread', 0)
    jam = kwargs.get('jam', 0)

    accessories = {
        'milk': milk,
        'egg': egg,
        'biscuit': biscuit,
        'bread': bread,
        'jam': jam
    }

    for item, cost in accessories.items():
        item = str(item).capitalize()
        print(f"{item} = ₹{cost}/-")

    try:
        sum_accessories = milk + egg + biscuit + bread + jam
    except Exception:
        print(f"Please provide valid numeric values for the accessories. String values are not allowed.")
        return 0
    return f"Total Invoice Amount = ₹{sum_accessories}/-"
```

**Use Case 7: local vs global**

Try the exercise given in the portal “Variable Scope (Local vs Global)“

```python
global_a = 10

def variable_scope():
    local_a = global_a
    return local_a

print("global_a =", global_a)
print("variable_scope(): global_a =", variable_scope())
print("global_a =", global_a)
```

```python
global_a = 10

def variable_scope():
    local_a = 20
    global_a = local_a
    return global_a

print("global_a =", global_a)
print("variable_scope(): global_a =", variable_scope())
print("global_a =", global_a)
```

```python
global_a = 10

def variable_scope():
    local_a = 20
    global global_a   # Assigning global scope to local variable
    global_a = local_a
    return global_a

print("global_a =", global_a)
print("variable_scope(): global_a =", variable_scope())
print("global_a =", global_a)
```
