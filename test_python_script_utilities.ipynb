```python
try:
    a = 1 / 0
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
```

```python
# print with separator
print("Hello", "World", "Python", sep="-")
print("Hello", "World", "Python", sep=" | ")
print("Hello", "World", "Python", sep="\n")  # New line separator
print("Hello", "World", "Python", sep="\t")  # Tab separator
```

```python
# print with end parameter
print("Hello", end=" ")
print("World", end=" ")
```
